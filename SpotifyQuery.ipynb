#Upload Hadoop Output into MongoDB

```python
SampleDb = client.ArtistInfo
user = SampleDb.Artists
```

```python
#Upload our Hadoop output into the MongoDB via Google Colab
with open("/content/drive/MyDrive/BigData/Artist_output_part-00000") as f:
  for line in f:
    cols = line.split()
    length = len(cols)
    name = ""
    num = 0

    for i in range(length):
      if (i == length-1):
        num += int(cols[i])
      else:
        name += " "+cols[i]

    #insert each artist
    user.insert_one({
        'name' : name.strip(),
        'amount' : num
    })
```

#Using Query to receive the meaningful data

## For example, this is for finding Top 10 aritist who has the most songs.

```python

Top10 = list(user.aggregate([
    {
    "$sort":{"amount":-1}
    },
    {
    "$limit":10
    }
]))

for e in Top10:
  print(e)
```
